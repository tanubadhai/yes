```python
class BaseSalary:
  def __init__(self,base_salary, bonus_rate=0.1, symbol="$"):
    self.base_salary = base_salary
    self.bonus_rate = bonus_rate
    self.symbol = symbol
    self.total_salary = base_salary * (1+bonus_rate)
    self.bonus = self.total_salary - self.base_salary
  def __repr__(self):
    return f'{self.symbol}{self.base_salary:,}'
  def show_salary(self):
    return f'{self.symbol}{self.total_salary:,}'
  def show_bonus(self):
    return f'{self.symbol}{self.bonus:,}'
```

```python
salary = BaseSalary(100000)
salary.symbol
```

```python
salary.show_bonus()

```

```python
salary.show_salary()
```
